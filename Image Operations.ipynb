```python
import cv2
from google.colab.patches import cv2_imshow
import numpy as np

# ----- Image Operations -----
img = cv2.imread('/content/F1.jpg')  # File name only (if uploaded to /content/)
if img is None:
    print(" Image not found! Check filename or upload again.")
else:
    cv2_imshow(img)

    # Negative
    negative = 255 - img

    # Brightness (+50)
    brightness = cv2.convertScaleAbs(img, alpha=1, beta=50)

    # Contrast (alpha=1.5)
    contrast = cv2.convertScaleAbs(img, alpha=1.5, beta=0)

    # Combine all horizontally
    combined = np.hstack((img, negative, brightness, contrast))
    cv2_imshow(combined)

    cv2.imwrite('/content/image_operations_combined.jpg', combined)
```

```python
cap = cv2.VideoCapture('12469452_2160_3840_30fps.mp4')
while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break
    cv2_imshow(frame)
    if cv2.waitKey(1) & 0xFF == 27:
        break
cap.release()
```
