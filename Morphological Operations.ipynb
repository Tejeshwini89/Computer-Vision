```python
import cv2
import numpy as np
from matplotlib import pyplot as plt

# Step 1: Load image
img = cv2.imread('/content/img1.jpg')  # Change this to your image path
if img is None:
    raise ValueError("Image not found! Please check the file path.")

# Step 2: Convert to grayscale
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Step 3: Threshold (binarization)
_, binary = cv2.threshold(gray, 127, 255, cv2.THRESH_BINARY)

# Step 4: Define structuring element (kernel)
kernel = cv2.getStructuringElement(cv2.MORPH_RECT, (5, 5))

# Step 5: Apply morphological operations
erosion = cv2.erode(binary, kernel, iterations=1)
dilation = cv2.dilate(binary, kernel, iterations=1)
opening = cv2.morphologyEx(binary, cv2.MORPH_OPEN, kernel)
closing = cv2.morphologyEx(binary, cv2.MORPH_CLOSE, kernel)

# Step 6: Visualization
titles = [
    'Original Image',
    'Grayscale',
    'Binary',
    'Erosion',
    'Dilation',
    'Opening',
    'Closing'
]
images = [img, gray, binary, erosion, dilation, opening, closing]

plt.figure(figsize=(10, 8))
for i in range(len(images)):
    plt.subplot(3, 3, i + 1)
    if len(images[i].shape) == 2:
        plt.imshow(images[i], cmap='gray')
    else:
        plt.imshow(cv2.cvtColor(images[i], cv2.COLOR_BGR2RGB))
    plt.title(titles[i])
    plt.axis('off')

# Visualize the kernel separately
plt.subplot(3, 3, 8)
plt.imshow(kernel, cmap='gray')
plt.title('Kernel (Structuring Element)')
plt.axis('off')

plt.tight_layout()
plt.show()
```
